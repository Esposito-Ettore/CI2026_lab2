```python
from random import seed, random, choice, sample
from math import inf

seed(42)
N = 10
```

```python
objects = frozenset(i for i in range(N))
sets = tuple(frozenset(sample(range(N), k=i+1)) for i in range(N))
costs = tuple(10 * random() + (i + random())**2 for i in range(N))
```

```python
def fitness(sol: list[bool]) -> float:
    try:
        # I get the objects contained in the subsets selected by the solution
        selected_sets = frozenset.union(*(sets[i] for i in range(N) if sol[i]))
    except:
        return -inf # in case sol contains all false the union fails

    # If the objects are all included in the union then the fitness is the negative of the sum of costs in order to aim for lowest cost
    # If not all objects are included, then the solution is rejected by default
    if len(selected_sets) == N:
        return -sum(costs[i] for i in range(N) if sol[i])
    else:
        return -inf

def tweak(sol: list[bool]) -> list[bool]:
    new_sol = sol[:]
    # 50% of the time I simply invert the selection of a specific subset
    # 50% of the time I swap between a selected and not-selected subsets
    if choice([True, False]) or sum(sol) == N:
        pos = choice(range(N))
        new_sol[pos] = not new_sol[pos]
    else:
        pos = choice([i for i in range(N) if new_sol[i]])
        pos2 = choice([i for i in range(N) if not new_sol[i]])
        new_sol[pos] = False
        new_sol[pos2] = True
        
        
    return new_sol
```

```python
cur_sol = [True for _ in range(N)]
cur_fitness = fitness(cur_sol)
print(sum(cur_sol))
print(cur_fitness)
i = 0
while True:
    i += 1
    new_sol = tweak(cur_sol)
    new_fitness = fitness(new_sol)
    if new_fitness > cur_fitness:
        cur_sol = new_sol
        cur_fitness = new_fitness
        print(f"Iteration {i}: new solution fitness {cur_fitness}")
        print(cur_sol)
```
